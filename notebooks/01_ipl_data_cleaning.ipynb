```python
import pandas as pd
import numpy as np
```

1. Load Data

```python
matches = pd.read_csv(r"C:\Users\khera\OneDrive\Desktop\coding study material\IPL Complete Dataset (2008 - 2024)\matches.csv")
deliveries = pd.read_csv(r"C:\Users\khera\OneDrive\Desktop\coding study material\IPL Complete Dataset (2008 - 2024)\deliveries.csv")
```

2. Standardize Franchise Names 

```python
team_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiant': 'Rising Pune Supergiants',
    'Deccan Chargers': 'Sunrisers Hyderabad',
    'Bangalore': 'Royal Challengers Bengaluru'
}
team_cols = ['team1', 'team2', 'winner', 'toss_winner']
for col in team_cols:
    if col in matches.columns:
        matches[col] = matches[col].replace(team_mapping)

if 'batting_team' in deliveries.columns:
    deliveries['batting_team'] = deliveries['batting_team'].replace(team_mapping)
if 'bowling_team' in deliveries.columns:
    deliveries['bowling_team'] = deliveries['bowling_team'].replace(team_mapping)
```

3. Handle Missing Values

```python
matches['winner'] = matches['winner'].fillna('No Result')
matches['city'] = matches['city'].fillna('Unknown')
```

4. Export Cleaned CSVs

```python
matches.to_csv(r'C:\Users\khera\OneDrive\Desktop\coding study material\IPL Complete Dataset (2008 - 2024)\cleaned_matches.csv', index=False)
deliveries.to_csv(r'C:\Users\khera\OneDrive\Desktop\coding study material\IPL Complete Dataset (2008 - 2024)\cleaned_deliveries.csv', index=False)
print("Data Cleaning Complete!")
```
